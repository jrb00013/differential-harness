# VOH Spin Breakeven — Milestone #3 Graph

Spin vs no-spin: find ω where `P_net(ω) > P_net(0)`.

Spec: `docs/VOH_PHYSICS.md`

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

from simulation.constants import TAU_SPIN_DEFAULT_NM, rpm_from_omega
from simulation.vortex_osmotic_hydro import breakeven_omega, sweep_omega, voh_state
```

```python
sweep = sweep_omega(omega_values=list(np.linspace(0, 150, 31)))
flat = voh_state(omega_rad_s=0.0, delta_h_z_m=0.0)
be = breakeven_omega()
print('Flat P_net:', flat.P_net_W, 'W')
print('Breakeven:', be)
```

```python
rpm = [p['rpm'] for p in sweep]
P_net = [p['P_net_W'] for p in sweep]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(rpm, P_net, 'o-', color='#2b6cb0', lw=2, label='VOH P_net(ω)')
ax.axhline(flat.P_net_W, color='crimson', ls='--', label=f'Flat baseline ({flat.P_net_W:.2f} W)')
if be:
    ax.axvline(be['rpm'], color='green', ls=':', label=f"Breakeven {be['rpm']:.0f} RPM")
ax.set_xlabel('Spin rate (RPM)')
ax.set_ylabel('P_net (W)')
ax.set_title('VOH / Z-Hydro: spin vs no-spin')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
out = ROOT / 'exports' / 'figures' / 'fig16_voh_omega_nb.png'
out.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(out, dpi=200)
print('Saved', out)
plt.show()
```
